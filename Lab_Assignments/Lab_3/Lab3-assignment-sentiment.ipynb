## Sentence Analysis

Overall notes about VADER: The VADER library is a very powerful and popular lexicon-based sentiment analysis tool
used in many areas of natural language processing. VADER first breaks down the text into individual words(tokens),
then each word(token) is assigned a sentiment score based on the "vader_lexicon.txt" file, identifying each word
as either negative, neutral or positive. After all of this is done, VADER computes a 'compound' value which is the
overall sentiment of the text. What makes VADER so powerful and popular is its ability to identify the intensity of
sentiment; for example, capitalization, punctuation, and reinforcing words (modifiers) have a large impact on the
sentiment of the overall text.

Based on the notes about VADER, let's analyze the above sentences:

* Sentence 1: For this sentence the compound score seems reasonable, looking at each word in the sentence ['I',
'love', 'apples'] the word 'I' is mostly neutral, the word 'apples' does not exist within the lexicon file so is
most likely assumed neutral as well, however; the word 'love' is in the lexicon file (and has an individual mean
score of 3.2), accompanied by the fact that it is an intensifier for the neutral word 'I' the compound score makes
sense here.

* Sentence 2: For this sentence the compound score seems more interesting, looking at each word again the only
difference in this sentence is the addition of the word "don't," which negates the first sentence. Although the
sentiment value for somebody writing 'I love apples' and 'I don't love apples' is the same for humans, VADER seems
more unsure when the sentence is negated and therefore gives a greater 'neu' value. This shows us that common human
text which includes jokes, irony, sarcasm, and negations will be difficult for VADER to understand due to its
rule-based nature.

* Sentence 3: For this sentence the compound score makes a lot of sense, it is the same as the first sentence
except for the addition of ':-)' at the end, in the 'vader_lexicon.txt' file the smiley face has a mean score of 1.3, and while it is not a hugely large score it does have some form of intensity, therefore, when coupled with
'love' brings the compound score from about 0.64 to 0.76 which is a significant difference.

* Sentence 4: For this sentence the compound score does not make a lot of sense, for humans such a sentence would
have a significantly larger negative score than it does for VADER (depending on the context, of course). This
sentence contains the word 'ruins' at the end (with a mean score of -1.9), and although there is heavy intensity
projected onto the word 'houses,' VADER seems to not take this as much into account as humans would. This is another
example of where complex language structures cannot be fully captured via lexicon(rule)-based systems.

* Sentence 5: For this sentence the compound does make sense, considering the explanation regarding sentence 4.
Following the previous explanation, although a human would interpret this sentence more neutrally, the negation
of the word 'ruins' and the inclusion of the word 'certainly' before the negation (with an individual mean score of
1.4). While the mean score for 'certainly' is not high because VADER struggles with the negation of the negative
word 'ruins' and 'certainly' is a positive word (which can also be considered a slight intensifier), the compound
score reflects these elements in the sentence.

* Sentence 6: For this sentence the compound score is very strange, for any human this sentence is purely
informational, there is no emotion or response attached to this sentence without further context. Because
the word 'lies' is present within the sentence (with an individual mean score of -1.8) and intensifies both 'chair'
and 'garden,' the compound score reflects the intensity projected on the overall sentence.

* Sentence 7: Lastly, for this sentence the compound score is also a bit skewed, but less than sentence 6. Once
again, this sentence is purely informational and meant to convey that a house is similar to other houses, although
we are unsure of which qualities are compared, this is generally a neutral sentence. Having said that, we do also
understand that because of the word 'like' (which has an individual mean score of 1.5), and the intensity
amplification on the word 'house,' that the compound score is slightly raised.

````python
def vader_output_to_label(vader_output):
    """
    map vader output e.g.,
    {'neg': 0.0, 'neu': 0.0, 'pos': 1.0, 'compound': 0.4215}
    to one of the following values:
    a) positive float -> 'positive'
    b) 0.0 -> 'neutral'
    c) negative float -> 'negative'
    
    :param dict vader_output: output dict from vader
    
    :rtype: str
    :return: 'negative' | 'neutral' | 'positive'
    """
    compound = vader_output['compound']
    
    if compound < 0:
        return 'negative'
    elif compound == 0.0:
        return 'neutral'
    elif compound > 0.0:
        return 'positive'
    
assert vader_output_to_label( {'neg': 0.0, 'neu': 0.0, 'pos': 1.0, 'compound': 0.0}) == 'neutral'
assert vader_output_to_label( {'neg': 0.0, 'neu': 0.0, 'pos': 1.0, 'compound': 0.01}) == 'positive'
assert vader_output_to_label( {'neg': 0.0, 'neu': 0.0, 'pos': 1.0, 'compound': -0.01}) == 'negative'
````

````python
# Your code here
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import spacy

vader_model = SentimentIntensityAnalyzer()
nlp = spacy.load('en_core_web_sm')
all_text = my_tweets

def run_vader(textual_unit, 
              lemmatize=False, 
              parts_of_speech_to_consider=None,
              verbose=0):
    """
    Run VADER on a sentence from spacy
    
    :param str textual unit: a textual unit, e.g., sentence, sentences (one string)
    (by looping over doc.sents)
    :param bool lemmatize: If True, provide lemmas to VADER instead of words
    :param set parts_of_speech_to_consider:
    -None or empty set: all parts of speech are provided
    -non-empty set: only these parts of speech are considered.
    :param int verbose: if set to 1, information is printed
    about input and output
    
    :rtype: dict
    :return: vader output dict
    """
    doc = nlp(textual_unit)
        
    input_to_vader = []

    for sent in doc.sents:
        for token in sent:

            to_add = token.text

            if lemmatize:
                to_add = token.lemma_

                if to_add == '-PRON-': 
                    to_add = token.text

            if parts_of_speech_to_consider:
                if token.pos_ in parts_of_speech_to_consider:
                    input_to_vader.append(to_add) 
            else:
                input_to_vader.append(to_add)

    scores = vader_model.polarity_scores(' '.join(input_to_vader))
    
    if verbose >= 1:
        print()
        print('INPUT SENTENCE', sent)
        print('INPUT TO VADER', input_to_vader)
        print('VADER OUTPUT', scores)

    return scores
````
