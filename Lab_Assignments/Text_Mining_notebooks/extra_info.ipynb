## Cosine similarity

cosine(v,w) = (v x w / |v| x |w|) =>

=> (sum from i=1 to N (V_i x W_i) / (sqrt(sum from i=1 to N (V_i^2)) x sqrt(sum from i=1 to N (W_i^2))))

#### Example

cos(cherry,information) = (442 * 5 + 8 * 3982 + 2 * 3325) / sqrt(442^2 + 8^2 + 2^2) * sqrt(5^2 + 3982^2 + 3325^2) = 0.018

cos(digital,information) = (5 * 5 + 1683 * 3982 + 1670 * 3325) / sqrt(5^2 + 1683^2 + 1670^2) * sqrt(5^2 + 3982^2 + 3325^2) = 0.996
